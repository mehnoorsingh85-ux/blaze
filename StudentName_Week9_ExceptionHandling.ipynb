# Task 1 – Identify the errors
Run each of the three provided examples and observe the error.

### Example 1

```python
# Example 1
len(5)
```

**Error type: `TypeError`**

**Why it occurs:** `len()` only works on objects that have a length, such as strings, lists, tuples and dictionaries. `5` is an integer, which has no length, so Python raises a `TypeError`.

### Example 2

```python
# Example 2
def add(a, b):
    return a + b

add('hello', 5)
```

**Error type: `TypeError`**

**Why it occurs:** The `+` operator can add two numbers or join two strings, but it cannot combine a string (`'hello'`) with an integer (`5`). Python cannot mix the two types, so it raises a `TypeError`.

### Example 3

```python
# Example 3 - OverflowError
import math
x = math.exp(1000)
```

**Error type: `OverflowError`**

**Why it occurs:** `math.exp(1000)` calculates e^1000, which is far larger than the biggest floating-point number Python can store (about 1.8 × 10^308). The result is out of range, so Python raises an `OverflowError`.

---
# Task 2 & Task 3 – Handle the errors and test the solution
Each program below is modified with a `try...except` block that catches the correct exception, stops the program from crashing, and shows a clear message to the user.

### Example 1 – handled (`TypeError`)

```python
# Example 1 - handled with try...except
try:
    result = len(5)
    print("Length:", result)
except TypeError:
    print("Error: len() cannot be used on a number. Please provide a string, list, or other sequence.")

print("Program continues running...")
```

### Example 2 – handled (`TypeError`)

```python
# Example 2 - handled with try...except
def add(a, b):
    return a + b

try:
    result = add('hello', 5)
    print("Result:", result)
except TypeError:
    print("Invalid input. Please enter the correct type of value (both numbers or both strings).")

print("Program continues running...")

# Test with valid input to show the function still works normally
try:
    print("Valid test:", add(10, 5))
    print("Valid test:", add('hello', ' world'))
except TypeError:
    print("Invalid input. Please enter the correct type of value.")
```

### Example 3 – handled (`OverflowError`)

```python
# Example 3 - handled with try...except
import math

try:
    x = math.exp(1000)
    print("Result:", x)
except OverflowError:
    print("Error: The number is too large to calculate. Please use a smaller value.")

print("Program continues running...")

# Test with a valid value to show it works normally
try:
    print("Valid test:", math.exp(10))
except OverflowError:
    print("Error: The number is too large to calculate. Please use a smaller value.")
```
